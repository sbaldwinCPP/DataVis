```python
import os
import polars as pl

file_path = os.path.join(
    os.path.dirname(os.getcwd()),
    "src",
    "data",
    "nyc-tmy-2023.csv",
)

df = pl.read_csv(file_path, skip_rows=2)
df.group_by(["Year", "Month", "Day"], maintain_order=True).agg(
    pl.col("Temperature").max()
)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.scatter(
    x=df["Wind Direction"],
    y=df["Temperature"],
    # c=df["Wind Speed"],
    c=df["Month"],
    cmap="twilight_shifted",
    # cmap="jet",
    marker=".",
)

plt.show()
```

```python
fig, ax = plt.subplots()

ax.hist(
    x=df["Wind Direction"],
    bins=16,
)

plt.show()
```
